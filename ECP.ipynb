```python
import rsa
import hashlib
```

```python
plain_text = input('Введите Ваше сообщение: ')
```

```python
text_for_check = input('Введите Ваше сообщение для проверки (либо подмены🤔): ').encode()
```

```python
pubkey, privkey = rsa.newkeys(1024)
```

```python
hashed_msg = hashlib.sha256(plain_text.encode()).digest()  
```

```python
signature = rsa.sign(hashed_msg, privkey, 'SHA-256')
```

```python
hash_for_check = hashlib.sha256(text_for_check).digest()
```

```python
try:
    rsa.verify(hash_for_check, signature, pubkey)
    print(f'Сообщение: \n"{text_for_check.decode('utf-8')}"\nпередалось правильно')
except VerificationError:
    print('Нас где-то наебывают')
```
